```python
import matplotlib.pyplot as plt

df['type'].value_counts().plot(kind='bar')

plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Number of Titles')
plt.show()
```

This is our first data visualization and that containg in bar chart.

```python
df['release_year'].value_counts().sort_index()
```

Here python gives the release year according to the number of movies.

```python
year_counts = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(12, 5))
plt.plot(year_counts.index, year_counts.values)

plt.title('Netflix Titles by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(True)
plt.show()
```

This is our Line chart and gives when netflix content increased, when it decreased, which year had the most titles.

```python
country_data = df['country'].dropna().str.split(', ').explode()

country_counts = country_data.value_counts()

country_counts.head(10)
```

First we have data like comma seperated and now we have data like column wise and now python will easily counts them.

```python
top_countries = country_counts.head(10)

plt.figure(figsize=(10, 6))
top_countries.plot(kind='bar')

plt.title('Top 10 Countries by Netflix Titles')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

This is the 3rd visualization chat.

```python
type_year = df.groupby(['release_year', 'type']).size().unstack(fill_value=0)

type_year.tail(10)
```

It shows the last 10 years in the table.
groupby() gives the similar things together, size() counts them and unstack() reads the table.

```python
plt.figure(figsize=(12, 6))

plt.plot(type_year.index, type_year['Movie'], label='Movies')
plt.plot(type_year.index, type_year['TV Show'], label='TV Shows')

plt.title('Movies vs TV Shows by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

This chart helps us to compare Movies and TV shows across the years.
This chart helps us to look for trends.

```python
genre_data = df['listed_in'].dropna().str.split(', ').explode()

genre_counts = genre_data.value_counts()

genre_counts.head(10)
```

Movie A can belongs to multiple categories. So we need to seperate them before counting.

```python
top_genres = genre_counts.head(10)

plt.figure(figsize=(10, 6))
top_genres.plot(kind='bar')

plt.title('Top 10 Netflix Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

These are the most frequently listed genres among Netflix titles.

```python
director_counts = df['director'].dropna().value_counts()

director_counts.head(10)
```

dropna() ignore rows where the director is missing. value_counts() counts how many times each director appears.

```python
top_directors = director_counts.head(10)

plt.figure(figsize=(10, 6))
top_directors.plot(kind='bar')

plt.title('Top 10 Directors on Netflix')
plt.xlabel('Director')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

df[['title', 'date_added']].head(10)
```

```python
df['add_year'] = df['date_added'].dt.year

df[['title', 'date_added', 'add_year']].head(10)
```

```python
add_year_counts = df['add_year'].value_counts().sort_index()

add_year_counts
```

```python
plt.figure(figsize=(10, 5))

add_year_counts.plot(kind='bar')

plt.title('Netflix Titles Added by Year')
plt.xlabel('Year Added to Netflix')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

This is different from earlier charts.
Release year: when was movie/show released.
Add year: when was it added to netflix.

```python
import sqlite3

DB_PATH = '/content/drive/MyDrive/Data_Analytics_Project/Netflix_Content_Analytics/netflix.db'

conn = sqlite3.connect(DB_PATH)

print("Database connected successfully!")
```

```python
result = conn.execute("""
SELECT type, COUNT(*) AS total_titles
FROM netflix
GROUP BY type;
""")

for row in result:
    print(row)
```

```python
result = conn.execute("""
SELECT country, COUNT(*) AS total_titles
FROM netflix
WHERE country IS NOT NULL
GROUP BY country
ORDER BY total_titles DESC
LIMIT 10;
""")

for row in result:
    print(row)
```

```python
result = conn.execute("""
SELECT rating, COUNT(*) AS total_titles
FROM netflix
WHERE rating IS NOT NULL
GROUP BY rating
ORDER BY total_titles DESC;
""")

for row in result:
    print(row)
```

MA - Mature audience
PG - Parent Guidence
G - General Audience
R - Restricted
Y - Young Children

```python
result = conn.execute("""
SELECT type, rating, COUNT(*) AS total_titles
FROM netflix
WHERE rating IS NOT NULL
GROUP BY type, rating
ORDER BY type, total_titles DESC;
""")

for row in result:
    print(row)
```

```python
result = conn.execute("""
SELECT release_year, COUNT(*) AS total_titles
FROM netflix
WHERE release_year IS NOT NULL
GROUP BY release_year
ORDER BY release_year DESC
LIMIT 15;
""")

for row in result:
    print(row)
```
